# Welcome to the BRH Workspace with the Workspace Software Library

**This is your personal workspace. No one else can access the data or files here.**

You can learn more about using the BRH Workspace in the [BRH Documentation](https://uc-cdis.github.io/BRH-documentation/).

**The `pd` folder (find it in the panel on the left) is your persistent drive:**

* Use this folder to store files (notebooks, data files, etc) that you want to use again later. The files you save here will still be available when you come back after terminating your workspace session.
* **Any files you create or add outside of this folder will be lost if they are not moved to the `pd` before your workspace session terminates**.
* This image has a `pd` with a storage capacity limit of 10Gi

## Get started with Data Exploration

Open a new "Launcher" tab by clicking the `+` next to the `brh-welcome-wsl.ipynb` tab at the top of this document.

From the Launcher tab, you can:  
* open a new, empty Jupyter notebook;  
* open a new code console or terminal window; or,  
* create several types of files (text, Markdown, Python).  

For notebooks or files \- **remember to move the file into the `pd` directory if you want to access them in a later workspace session.**

### Download files from the BRH Discovery page to the Workspace  

[Learn how to download data files through the BRH Discovery page](https://uc-cdis.github.io/BRH-documentation/11-downloading_data_files/#download-data-files-into-a-workspace-with-the-python-sdk) in our documentation. 

When you download files from the [BRH Discovery Page](https://brh.data-commons.org/discovery) to the workspace, these will be placed in the folder `brh.data-commons.org` in the `data` folder. These will not persist beyond the current workspace session if they stay here. Move these files to the `pd` directory if you want to access them again after you terminate your workspace session.  

### Manage software in the workspace  

The BRH Workspace comes with many pre-installed software packages. You can view all installed software by opening a terminal window and using the command `pip list`.    

#### Using Pre-installed Software in the workspace  

The BRH Workspace Software Library contains many pre-installed packages to make data analysis easier. To view pre-installed packages, click the gear wheel icon in the left sidebar.   

* Packages that appear under “Loaded Modules” are ready to import (or can be unloaded, if desired).  
* Packages that appear under “Available Modules” can be loaded.  

After loading and/or unloading a package, you must go to the `Kernel menu` at the top navigation panel and select `Shut Down Kernel` and then `Reconnect to Kernel` for the changes to take effect.  

#### Installing additional software in the workspace  

If there are other software tools you want to use in the workspace, you can install them using `pip install` (Python) or `CRAN` (R; not available in this workspace yet). If you have a requirements text, you can upload your requirements text and conveniently install all of them with `pip install -r requirements.txt`.  

### Upload files from local storage    

You can [upload files to the workspace](https://uc-cdis.github.io/BRH-documentation/09-workspace_page/#upload-save-and-download-filesnotebooks) by clicking the upload arrow icon in the upper left corner.

If the locally-stored files you wish to analyze are large in number and/or size, you may need to zip them before uploading to a workspace. Once in a workspace, files can be unzipped using the python library [zipfile](https://docs.python.org/3/library/zipfile.html).

### Storage in the Workspace

The Jupyter Lab with Python and R workspace image is currently sized for small analyses and testing. The maximum storage in the `pd` for this image (including any notebooks or tools installed) is 10Gi. If you need a larger image for your analyses, please reach out to the BRH Data Platform support team at [brhsupport@gen3.org](mailto:brhsupport@gen3.org) to discuss the possibility of a larger image.
